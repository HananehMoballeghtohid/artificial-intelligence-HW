### Facts

```python
alice = Person("Alice")
dr_smith = Person("DrSmith")
lab2 = Room("Lab2")
```

```python
facts = [
    Student(alice), # Alice is a student
    HasID(alice), # Alice has a valid ID card
    Professor(dr_smith), # DrSmith is a professor
    Mentor(dr_smith, alice), # DrSmith is Alice's mentor
    InRoom(dr_smith, lab2), # DrSmith is physically in Lab2
    IsSecureLab(lab2), # Lab2 is a high -security lab
    DoorOpen(lab2) # The door to Lab2 is currently OPEN
]
```

```python
for f in facts:
    print(f)
```

```python
agent = HybridAccessAgent()
print(agent.decide_with_report(alice, lab2, facts))
```

```python
facts.remove(DoorOpen(lab2))
print(agent.decide_with_report(alice, lab2, facts))
```
